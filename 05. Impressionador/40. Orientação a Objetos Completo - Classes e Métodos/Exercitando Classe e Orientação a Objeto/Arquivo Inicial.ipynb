### Exemplo simples

```python
vendedor = "Paulo"
vendas = 1000
meta = 500

if vendas > meta:
    bonus = 20
else:
    bonus = 0
```

```python
class Vendedor:

    def __init__(self, nome):
        self.nome = nome
        self.vendas = 0
        self.meta = 500
        self.bonus = 0

    def vendeu(self, quantidade_vedas):
        self.vendas = quantidade_vedas
        self.calcular_bonus() # Calculamos o bonus altomaticamente, assim q a venda é cadastrada.

    def calcular_bonus(self):
        if self.vendas > self.meta:
            self.bonus = 50
        else:
            self.bonus = 0

vendedor_1 = Vendedor("Paulo")
vendedor_1.vendeu(1500)

vendedor_2 = Vendedor("Lucas")
vendedor_2.vendeu(300)

print(vendedor_1.nome, "Vendas:", vendedor_1.vendas, "Bonus:", vendedor_1.bonus)
print(vendedor_2.nome, "Vendas:", vendedor_2.vendas, "Bonus:", vendedor_2.bonus)
```

```python
import pyautogui
import pyperclip
import time

# abrindo o chrome
pyautogui.PAUSE = 1
pyautogui.press("win")
pyautogui.write("chrome")
pyautogui.press("enter")

# entrando no site da hashtag
link = "https://www.hashtagtreinamentos.com/blog"
pyperclip.copy(link)
pyautogui.hotkey("ctrl", "v")
pyautogui.press("enter")

# aguardar
time.sleep(3)

# clicando no campo de busca
pyautogui.click(x=1508, y=585)

# pesquisando campo de busca
texto = "classe"
pyperclip.copy(texto)
pyautogui.hotkey("ctrl", "v")
pyautogui.press("enter")

# aguardar
time.sleep(3)

# clicar na imagem
pyautogui.click(x=749, y=699)

# aguardar
time.sleep(3)

# extrair o link
pyautogui.click(x=390, y=789, button="right")
pyautogui.press("up")
pyautogui.press("up")
pyautogui.press("enter")

# printar o texto copiado
texto = pyperclip.paste()
print(texto)
```
